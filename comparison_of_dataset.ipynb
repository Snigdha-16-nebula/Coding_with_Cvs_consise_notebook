```python
#for mcvs
import pandas as pd

recent_data_pos = pd.read_csv("lightcurves_grouped_r_clean.csv")
previous_data_pos = pd.read_csv("lightcurves_positives_r_band.csv")

print("recent_data_pos:", recent_data_pos.shape)
print("previous_data_pos:", previous_data_pos.shape)

print("Unique objects in recent_data_pos:",
      recent_data_pos["object"].nunique())

print("Unique objects in previous_data_pos:",
      previous_data_pos["object"].nunique())

print("Common objects:",
      len(set(recent_data_pos["object"]) & set(previous_data_pos["object"])))

print("Only in recent_data_pos:",
      len(set(recent_data_pos["object"]) - set(previous_data_pos["object"])))

print("Only in previous_data_pos:",
      len(set(previous_data_pos["object"]) - set(recent_data_pos["object"])))
```

```python
recent_data = set(recent_data_pos["object"].dropna().unique())
previous_data = set(previous_data_pos["object"].dropna().unique())

common = recent_data & previous_data
missing_from_recent_data_pos= previous_data - recent_data
missing_from_previous_data_pos = recent_data - previous_data
print("Common objects:", len(common))

print("\nObjects present in previous_data but missing from recent_data:")
print(sorted(missing_from_recent_data_pos))

print("\nObjects present in recent_data but missing from previous_data:")
print(sorted(missing_from_previous_data_pos))
```
